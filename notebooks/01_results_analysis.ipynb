# LoCoMo experiment results

Reads harness aggregates only (`aggregate/runs.parquet`, `aggregate/examples.parquet`). Does not call LLMs.

Source: `python -m src.memorybench aggregate configs/experiments/<name>.yaml` after QA (and optionally autorater) `_SUCCESS` markers exist.


```python
from pathlib import Path

import pandas as pd

EXPERIMENT = "locomo-poc"
ROOT = Path("..") if not Path("experiments").exists() else Path(".")
agg = ROOT / "experiments" / EXPERIMENT / "aggregate"

runs = pd.read_parquet(agg / "runs.parquet")
examples = pd.read_parquet(agg / "examples.parquet")
print("runs", len(runs), "examples", len(examples))
runs.head()
```

```python
if len(examples):
    print(
        examples.groupby(
            ["reader_display_name", "memory_method"], dropna=False
        )[["locomo_f1", "judge_score"]].mean()
    )
```
